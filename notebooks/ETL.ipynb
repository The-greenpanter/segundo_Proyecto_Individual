```python
#  Eliminar columnas irrelevantes (columnas Unnamed)
def eliminar_columnas_irrelevantes(df):
    df_cleaned = df.loc[:, ~df.columns.str.contains('^Unnamed')]
    return df_cleaned
```

```python

accVelLocSinRangos = eliminar_columnas_irrelevantes(accVelLocSinRangos)
velocidadPorProv = eliminar_columnas_irrelevantes(velocidadProv)
totalesAccesosPorVelocidad = eliminar_columnas_irrelevantes(accesosVelocidad)
accesosTecnologiaLocalidad = eliminar_columnas_irrelevantes(accesosTecnologiaLocalidad)
penetracionHogares = eliminar_columnas_irrelevantes(penetracionHogares)
penetracionPoblacion = eliminar_columnas_irrelevantes(penetracionPoblacion)
ingresos = eliminar_columnas_irrelevantes(ingresos)
```

```python
# Gestionar los valores faltantes
def gestionar_valores_faltantes(df):
    # Mostrar los valores faltantes en el DataFrame
    print(df.isnull().sum())
    
    # Eliminar filas con valores faltantes en columnas clave (comprobamos las columnas)
    if 'Año' in df.columns and 'Trimestre' in df.columns and 'Provincia' in df.columns:
        df_cleaned = df.dropna(subset=['Año', 'Trimestre', 'Provincia'], how='any')
    else:
        print("Columnas clave no encontradas, aplicando solo limpieza por medianas.")
        df_cleaned = df.copy()

    # Rellenar valores faltantes en columnas numéricas con la mediana (ejemplo)
    for col in df_cleaned.select_dtypes(include=['float64', 'int64']).columns:
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())
    
    return df_cleaned
```

```python

# Aplicar a los DataFrames
accVelLocSinRangos = gestionar_valores_faltantes(accVelLocSinRangos)
velocidadPorProv = gestionar_valores_faltantes(velocidadProv)
totalesAccesosPorVelocidad = gestionar_valores_faltantes(totalesAccesosPorVelocidad)
accesosTecnologiaLocalidad = gestionar_valores_faltantes(accesosTecnologiaLocalidad)
penetracionHogares = gestionar_valores_faltantes(penetracionHogares)
penetracionPoblacion = gestionar_valores_faltantes(penetracionPoblacion)
ingresos = gestionar_valores_faltantes(ingresos)
```
